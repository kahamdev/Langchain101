# **03. We load dotenv from initializations environmental setups .**

```python
import os
from dotenv import load_dotenv

load_dotenv()

print("OpenAI Key Loaded:", os.environ.get("OPENAI_API_KEY") is not None)
print("ANTHROPIC Key Loaded:", os.environ.get("ANTHROPIC_API_KEY") is not None)
print("GOOGLE GENAI Key Loaded:", os.environ.get("GOOGLE-GENAI_API_KEY") is not None)
```

# **04. We initiate the llm from api model provider.**

```python
"""
 we call the chatopenai from langchain openai library
"""
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(
    model = "gpt-4o-mini",
    temperature = 0.5,
    metadata = {"description":"This is openai model we use "}
)
```

# **05. We create custom tool to called by agent.**

```python
""" we create custom weather tool then we binding it with tool method """
from langchain.tools import tool
@tool
def get_custom_weather_tool(city:str)->str:
    """ this is weather tool docstings """
    return f"It is 28°C and sunny in {city}."
```

# **06. We create our agent here .**

```python
""" we call create_agent method  from langchain tool and also call InMemorySaver for memory reference from langgraph tool  """
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
agent = create_agent(
    model = llm,
    tools = [get_custom_weather_tool],
    system_prompt = "You are helpful AI assistant",
    #checkpointer = InMemorySaver ()
)
```

# **07. We test our agent for input task execution .**

```python
""" We call invoke method to test action from agent created ."""
response = agent.invoke(
    {"messages":[{"role":"user", "content":"What is the weather in Dar es Salaam? "}]}
)
```

# **08. We execute the agent task action to see output result.**

```python
""" we use print method to dsiplay results ."""
print("Agent Results :",response["messages"][-1].content)
```
